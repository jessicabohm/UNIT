```python
import numpy as np
import matplotlib.pyplot as plt
import SimpleITK as sitk
import os
import csv
import pandas as pd

import scipy.stats
from sklearn.linear_model import LinearRegression

import torch
from networks_update import Encoder, Decoder
import math
import random
```

```python
mouse_train = "../3D-CycleGan-Pytorch-MedImaging/Data_folder_train_2/train/images/"
mouse_test = "../3D-CycleGan-Pytorch-MedImaging/Data_folder_train_2/test/images/"

human_train = "../3D-CycleGan-Pytorch-MedImaging/Data_folder_train_2/train/labels/"
human_test = "../3D-CycleGan-Pytorch-MedImaging/Data_folder_train_2/test/labels/"

for path, save_folder in [(mouse_train, "mouse_train/"), (mouse_test ,"mouse_test/"), (human_train, "human_train/"), (human_test, "human_test/")]:
    for file in os.listdir(path):
        img_arr = sitk.GetArrayFromImage(sitk.ReadImage(path + file))
        for i in range(-5, 5):
            slice_idx = img_arr.shape[0] // 2 + i
            slice_img_arr = img_arr[slice_idx]

            # size = 3
            # plt.figure(figsize=(size, size))
            # plt.imshow(slice_img_arr, cmap="gray", alpha=1.0)
            # plt.show()
            
            slice_img = sitk.GetImageFromArray(slice_img_arr)
            sitk.WriteImage(slice_img, "./datasets/2d_larger/" + save_folder + "s_" + str(i) + "_" + file)
```

```python
# test
for path, save_folder in [(mouse_train, "mouse_train/"), (mouse_test ,"mouse_test/"), (human_train, "human_train/"), (human_test, "human_test/")]:
    for file in os.listdir(path)[:2]:
        for i in range(-5, 5):
            slice_img_arr = sitk.GetArrayFromImage(sitk.ReadImage("./datasets/2d_larger/" + save_folder + "s_" + str(i) + "_" + file))
            
            size = 3
            plt.figure(figsize=(size, size))
            plt.imshow(slice_img_arr, cmap="gray", alpha=1.0)
            plt.show()
```
